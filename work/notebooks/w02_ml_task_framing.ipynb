## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

**Task Type**: Classification.

**Why**: The goal is to predict whether a specific piece of content will experience a significant drop in search impressions (a "decline"). Since the outcome we are predicting is a binary "yes/no" (it declines or it does not), this is a classification problem.

```python
# No complex code needed here, just verifying our task type choice.
task_type = "Classification"
print(f"Task Type: {task_type}")
```

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

**Target**: The target is `is_declining_label` (1 if declining, 0 if not).

**Source**: This is an **observed outcome** in the data. The label comes from `trend_direction == "down"`, which is defined as a 20%+ drop in search impressions in the last 30 days compared to the previous 30 days (`trend_pct`). It is calculated based on actual historical performance.

```python
import pandas as pd
import warnings
warnings.filterwarnings('ignore')

# Load the starter data to examine the target
df = pd.read_csv('../../data/raw/content_refresh_anonymized.csv')
df["is_declining_label"] = (df["trend_direction"] == "down").astype(int)
print("Target Distribution:")
print(df["is_declining_label"].value_counts(normalize=True))
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*

**Success Metric**: Precision.

**Why**: The action we take based on this prediction is dedicating editor time to refresh the content. Editor time is expensive (cost of a wrong call is wasted hours). We want to be highly confident that when we flag a page as "declining," it actually needs a refresh. A high precision ensures that the content we queue for editors is truly decaying.

```python
# Example of setting a baseline for the metric
baseline_precision = df["is_declining_label"].mean()
print(f"Baseline Precision (if we guessed randomly): {baseline_precision:.1%}")
print("Our model needs to beat this significantly to justify the editor effort.")
```

## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

**Unit of Analysis**: One row = **one pseudonymized content item (page)**.

```python
# Displaying the unit of analysis
display_df = df[["content_id", "client_id", "content_type", "impressions_90d", "clicks_90d", "is_declining_label"]].head(3)
display(display_df)
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

**Why ML**: A fixed rule (e.g., "if clicks drop by 20%") is too rigid. Content decline depends on many tangled signals: the content's age, the intent (`informational` vs `transactional`), keyword competition, and historical traffic volume. A simple if-statement cannot weigh these shifting, interconnected factors accurately, whereas an ML model can find the nuanced patterns across these dimensions to predict decline more reliably.

```python
# Demonstrating how features overlap and make simple rules hard
summary = df.groupby("content_type")["is_declining_label"].mean()
print("Decline rate varies by content type, showing interactions:")
print(summary)
```
